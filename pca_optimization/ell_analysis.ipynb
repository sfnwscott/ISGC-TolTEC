```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1.anchored_artists import AnchoredSizeBar
import matplotlib.colors as cols
from astropy.io import fits
from astropy.wcs import WCS
from astropy.coordinates import SkyCoord
import astropy.units as u
from astropy.table import QTable
import time
```

# No Fruitloops

```python
ell_dat = pd.read_csv('../datasets/ell_cutoff_signal_table.csv').drop(columns=['Unnamed: 0'])
ell_dat = ell_dat[ell_dat['ell_cutoff']!=0.1]
```

```python
cmap = plt.colormaps['turbo']
color_indices = np.linspace(0.1, 0.9, ell_dat['ell_cutoff'].nunique())
colors = cmap(color_indices)
colors = dict(zip(ell_dat['ell_cutoff'],colors))

fig, ax = plt.subplots(figsize=(8, 5))
for ell, color in colors.items():
    subset = ell_dat[ell_dat['ell_cutoff'] == ell]
    subset = subset.groupby('N-eig')['S_nonorm'].agg(['mean','std']).reset_index()
    init_sig = subset.loc[subset['N-eig']==1,'mean'].values[0]
    ax.errorbar(x=subset['N-eig'], y=subset['mean'], yerr=subset['std'], capsize=3, color=color, label=ell)
ax.set_xlabel(r'$N_{eig}$')
ax.set_ylabel(r'Signal')
ax.legend(title=r"$\ell_{cutoff} (arcmin^{-1})$")
ax.grid(True, linestyle='--', alpha=0.6)
ax.set_xlim([-10,120])
ax.set_ylim([0,3e8])
ax.set_facecolor('#f7f7f7')
fig.tight_layout()
ax.set_title('Signal Cleaning Analysis')#\n(Un-Normalized)')
plt.show()
```
